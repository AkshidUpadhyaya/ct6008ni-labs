```python
%sql
-- query 1
SELECT *
FROM orders_2026
WHERE order_id = "f29770f8-b2d5-4f5b-b5af-fa97d642ac63"
```

```python
%sql
--query 2
SELECT date_format(order_purchase_timestamp, 'yyyy-MM') as order_month,
       order_status,
       count(*) as total_orders
FROM orders_2026
GROUP BY order_month, order_status
ORDER BY order_month, order_status;
```

```python
%sql
-- creating a copy table
create table orders_2026_copy as
SELECT *
FROM orders_2026;
```

```python
%sql
update orders_2026_copy
set order_status = "canceled"
where order_id = "f29770f8-b2d5-4f5b-b5af-fa97d642ac63"; 
```

```python
%sql
DESCRIBE HISTORY orders_2026_copy;
```
