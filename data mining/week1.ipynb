```python
%sql
SELECT COUNT(*) AS total_orders FROM workspace.default.orders_2026
```

```python
%sql
SELECT order_status, COUNT(*) AS orders
FROM orders_2026
GROUP BY order_status;
```

```python
%sql
SELECT month(order_purchase_timestamp) as order_month, count(*) AS orders
FROM orders_2026
GROUP BY order_month
ORDER BY order_month;
```

```python
%sql
SELECT AVG(datediff(order_delivered_customer_date, order_purchase_timestamp)) AS avg_delivery_days
FROM orders_2026
WHERE order_status = 'delivered';
```

```python
%sql
SELECT date(order_purchase_timestamp) AS order_date, COUNT(*) AS total_orders
FROM orders_2026
GROUP BY order_date
ORDER BY total_orders DESC
LIMIT 1;
```

```python
%sql
SELECT COUNT(*) AS late_orders
FROM orders_2026
WHERE order_status = 'delivered' AND order_delivered_customer_date > order_estimated_delivery_date;
```

```python
%sql
SELECT month(order_purchase_timestamp) AS order_month,
AVG(IF(order_status = 'canceled', 1, 0)) AS cancellation_rate
FROM orders_2026
GROUP BY order_month
ORDER BY cancellation_rate DESC
LIMIT 1;
```
